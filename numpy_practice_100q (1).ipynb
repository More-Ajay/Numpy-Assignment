**Q53.**  
Given `scores = np.array([85, 92, 78, 65, 88, 71, 95])`:
- Use `np.argsort` twice to assign **ranks** (rank 1 = highest score)
- The formula is: `ranks = np.argsort(np.argsort(-scores)) + 1`
- Print scores alongside their ranks

---

````python
scores = np.array([85, 92, 78, 65, 88, 71, 95])

ranks = np.argsort(np.argsort(-scores)) + 1  # Negate the scores so that larger scores come first: [-85, -92, -78, -65, -88, -71, -95]
                                             # First argsort gives the indices that would sort the scores in descending order: np.argsort(-scores) [6, 1, 4, 0, 2, 5, 3]
                                             # Applying argsort again converts that ordering into rank positions: np.argsort(np.argsort(-scores))[3, 1, 4, 6, 2, 5, 0]
                                             # Add 1 to make ranks start at 1 instead of 0: ranks = [4, 2, 5, 7, 3, 6, 1]
                                             # So the highest score (95) gets Rank 1, the next highest (92) gets Rank 2, and so on.
print('Scores:',scores)
print('Ranks:',ranks)

for score , rank in zip(scores , ranks):
  print(f"Score: {score} , Rank: {rank}")
````
